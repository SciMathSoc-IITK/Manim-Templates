  4. *Dual-Panel Decomposition*: Side-by-side comparison cards (e.g. infinite lattice vs. finite torsion).
  5. *Discrete / Modular Clock*: Residue chips, cyclic modular arrows, congruence equations.
  6. *2D Grid Enumeration*: Animated $N \times N$ matrix scanning, tick marks, and solution pop-ins.
  7. *Statistical Metric Cards*: $2 \times 2$ prime data grid with lagged entrance animations.
  8. *Hierarchical Assembly*: Token cards $\to$ local factors $\to$ global product / series.
  9. *Analytic Continuation & Roots*: Convergence interval $\to$ continuation arrow $\to$ zero-order root comparisons ($r=0, 1, 2$).

## 5. Pre-Flight Verification & Mathematical Syntax Check

Verifies Python syntax, compiles all embedded $\LaTeX$ expressions with `pdflatex`, and ensures segment durations match the audio timeline.

```python
from pathlib import Path
import ast, subprocess, tempfile

script_path = Path("/content/universal_manim_reel.py")
source = script_path.read_text(encoding="utf-8")
compile(source, str(script_path), "exec")
print("✓ Python syntax verification: PASSED")

# Verify LaTeX compilation of all static formulas
tree = ast.parse(source)
tex_strings = []
for node in ast.walk(tree):
    if isinstance(node, ast.Call) and isinstance(node.func, ast.Name):
        if node.func.id in ("math_text", "tex_label", "formula_banner", "footer_note"):
            if node.args and isinstance(node.args[0], ast.Constant) and isinstance(node.args[0].value, str):
                tex_strings.append(node.args[0].value)

with tempfile.TemporaryDirectory() as d:
    doc = "\\documentclass{article}\\usepackage{amsmath,amssymb}\\begin{document}\n"
    for s in tex_strings:
        doc += f"$\\displaystyle {s}$\\\\\n"
    doc += "\\end{document}"
    (Path(d) / "test.tex").write_text(doc)
    res = subprocess.run(["pdflatex", "-interaction=nonstopmode", "test.tex"], cwd=d, capture_output=True)
    if res.returncode == 0:
        print(f"✓ LaTeX validation: PASSED ({len(tex_strings)} expressions compiled cleanly)")
    else:
        print("⚠ LaTeX warnings encountered (check logs if symbols fail to render).")
```

## 8. Generate Scene-Synchronized AI Voiceover Track

Synthesizes neural speech segment by segment, automatically adjusts speaking rates, and pads remaining scene tails with silence to lock narration to animation cues.

```python
import asyncio
from pathlib import Path
import subprocess
import edge_tts
from IPython.display import Audio, display

voice_dir = Path("/content/voice_segments")
voice_dir.mkdir(parents=True, exist_ok=True)
RATE_STEPS = ["+0%", "+5%", "+10%", "+15%", "+20%", "+25%"]

def probe_dur(p: Path) -> float:
    cmd = ["ffprobe", "-v", "error", "-show_entries", "format=duration",
           "-of", "default=noprint_wrappers=1:nokey=1", str(p)]
    return float(subprocess.run(cmd, capture_output=True, text=True, check=True).stdout.strip())

async def generate_scene_speech():
    padded_wavs = []
    print(f"Generating neural voiceover using: {REEL_CONFIG['voice_name']}")
    for idx, (scene_id, slot, _, script_text) in enumerate(SCENE_PIPELINE, start=1):
        raw_mp3 = voice_dir / f"{idx:02d}_{scene_id}.mp3"
        chosen_rate = None
        for rate in RATE_STEPS:
            tts = edge_tts.Communicate(script_text, voice=REEL_CONFIG['voice_name'], rate=rate)
            await tts.save(str(raw_mp3))
            dur = probe_dur(raw_mp3)
            if dur <= slot - 0.20:
                chosen_rate = rate
                break
        if chosen_rate is None:
            chosen_rate = RATE_STEPS[-1]

        wav_path = voice_dir / f"{idx:02d}_{scene_id}.wav"
        subprocess.run([
            "ffmpeg", "-y", "-i", str(raw_mp3),
            "-af", f"aresample=48000,apad,atrim=duration={slot:.3f}",
            "-ac", "2", "-ar", "48000", "-c:a", "pcm_s16le", str(wav_path)
        ], check=True, capture_output=True)
        padded_wavs.append(wav_path)
        print(f"   • Scene {idx:02d} ({scene_id:20s}): slot={slot:5.1f}s | speech={dur:5.2f}s | rate={chosen_rate}")

    concat_txt = voice_dir / "concat.txt"
    concat_txt.write_text("\n".join(f"file '{p.as_posix()}'" for p in padded_wavs) + "\n")
    master_wav = Path("/content/Master_AI_Voiceover.wav")
    subprocess.run([
        "ffmpeg", "-y", "-f", "concat", "-safe", "0", "-i", str(concat_txt),
        "-c:a", "pcm_s16le", "-ar", "48000", "-ac", "2", str(master_wav)
    ], check=True, capture_output=True)
    return master_wav

master_wav = await generate_scene_speech()
print(f"\n✓ Master voiceover assembled: {master_wav} ({probe_dur(master_wav):.2f}s)")
display(Audio(str(master_wav)))
```

## 9. Audio Muxing & Master Video Export

Combines the 1080p silent video with the synchronized voiceover track and applies fade-in/fade-out curves.

```python
import subprocess
from pathlib import Path
from IPython.display import Video, display

final_export = Path("/content/SciMathSoc_Reel_FINAL_WITH_VOICEOVER.mp4")
hd_silent = Path("/content/Reel_HD_Master_Silent.mp4")
master_wav = Path("/content/Master_AI_Voiceover.wav")
dur = probe_dur(master_wav)

cmd = [
    "ffmpeg", "-y",
    "-i", str(hd_silent),
    "-i", str(master_wav),
    "-filter_complex", f"[1:a]aresample=48000,afade=t=in:st=0:d=0.20,afade=t=out:st={dur-0.8:.2f}:d=0.75[aout]",
    "-map", "0:v:0", "-map", "[aout]",
    "-c:v", "copy", "-c:a", "aac", "-b:a", "192k", "-shortest",
    "-movflags", "+faststart", str(final_export)
]

print("Muxing final video with synchronized voiceover...", flush=True)
res = subprocess.run(cmd, text=True, capture_output=True)
if res.returncode != 0:
    print(res.stderr[-2000:])
    raise RuntimeError("Audio muxing failed.")

print(f"✓ Master Reel exported successfully: {final_export} ({final_export.stat().st_size/(1024**2):.2f} MiB)")
display(Video(str(final_export), embed=True, width=420))
```

## 10. Generate Subtitles (.srt) & Social Media Captions

Generates a timestamped `.srt` subtitle file from the scene pipeline and social-media-ready captions with Unicode mathematical characters.

```python
from pathlib import Path

# Build SRT from pipeline
def format_srt_time(seconds: float) -> str:
    millis = int(round((seconds - int(seconds)) * 1000))
    s = int(seconds) % 60
    m = (int(seconds) // 60) % 60
    h = int(seconds) // 3600
    return f"{h:02d}:{m:02d}:{s:02d},{millis:03d}"

srt_entries = []
cur_time = 0.0
for idx, (_, slot, _, text) in enumerate(SCENE_PIPELINE, start=1):
    start_str = format_srt_time(cur_time)
    end_str = format_srt_time(cur_time + slot)
    srt_entries.append(f"{idx}\n{start_str} --> {end_str}\n{text}\n")
    cur_time += slot

srt_path = Path("/content/subtitles.srt")
srt_path.write_text("\n".join(srt_entries), encoding="utf-8")
print(f"✓ Subtitles generated: {srt_path} ({len(SCENE_PIPELINE)} cues)")
```
